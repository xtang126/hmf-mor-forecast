# MOR under incompleteness: a two-bin toy study
**How below affect the fits of a mass–observable relation (MOR), how it evolve with the increased factor of the losts and further to the cosmology**
1. Different threshold-based selection cut on observables
2. Mass dependent scatter and slope(tbd)
3. How much we know about the lost clusters
4. Consider halo mass function (hmf) and Eddington bias

Dependency: `numpy` + `scipy` + `matplotlib`. The setup mirrors the physical picture: two disjoint mass bins with intrinsic log-normal scatter, a single observable threshold that preferentially kills the low-mass, low-observable clusters.

## Physical setup

We assume a log-linear MOR
$$\ln \mathrm{Obs} = A + \alpha \, \ln(M / M_\mathrm{piv}) + \varepsilon, \qquad \varepsilon \sim \mathcal{N}(0, \sigma_{\ln})$$

with fiducial $A = 0$, $\alpha = 1.5$, $\sigma_{\ln} = 0.2$, $M_\mathrm{piv} = 3\times 10^{14}\,M_\odot/h$. This form is the standard log-normal MOR used throughout the cluster-cosmology literature ([Lima & Hu 2005](https://arxiv.org/abs/astro-ph/0503363); [Allen, Evrard & Mantz 2011](https://arxiv.org/abs/1103.4829)).

Clusters are drawn in two disjoint bins, uniform in $\log_{10}M$:
- Low: $M \in [5\times 10^{13}, 10^{14}]\,M_\odot/h$
- High: $M \in [5\times 10^{14}, 10^{15}]\,M_\odot/h$

Detection is a single threshold: a cluster survives iff $y = \ln \mathrm{Obs} > y_\mathrm{min}$. Because low-mass clusters have lower mean $y$, a single threshold **automatically** produces higher incompleteness in the low bin — no manual tuning of two separate loss fractions. Real X-ray or SZ surveys behave this way (see e.g. [Vikhlinin et al. 2009](https://arxiv.org/abs/0805.2207); [Mantz et al. 2010](https://arxiv.org/abs/0909.3098)).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy.optimize import minimize

# ---- Fiducial MOR parameters ----
A_TRUE     = 0.0            # intercept (choice of Obs units)
ALPHA_TRUE = 1.5            # slope
SIGMA_TRUE = 0.2            # intrinsic scatter in ln(Obs)
M_PIV      = 3e14           # pivot mass, M_sun/h

# ---- Mass bins (disjoint) ----
M_EDGES_LOW  = (5e13, 1e14)
M_EDGES_HIGH = (5e14, 1e15)

# ---- Sample size per bin ----
N_LOW  = 100
N_HIGH = 50

# Master RNG (a fresh RNG is used for each realisation inside the scan so
# results are reproducible per (threshold, realisation) index)
rng_master = np.random.default_rng(42)

# ---- Mass-dependent scatter parameters ----
# sigma_lnO(M) = SIGMA_TRUE * (M / M_piv) ** GAMMA_SCATTER
# GAMMA_SCATTER = 0    -> constant scatter (original behaviour, backward compatible)
# GAMMA_SCATTER < 0    -> LARGER scatter at low mass (physically motivated)
# GAMMA_SCATTER > 0    -> larger scatter at high mass
GAMMA_SCATTER = 0.0    # try 0.0, -0.3, -0.6 to see the effect grow
```

### Note on mass-dependent scatter

The default scatter model in this notebook is now mass-dependent:
$$\sigma_{\ln O}(M) = \sigma_0 \cdot (M / M_\mathrm{piv})^{\gamma_\mathrm{scatter}}$$

This is the standard parameterization for X-ray L–M and Y–M relations, following [Evrard et al. 2014](https://arxiv.org/abs/1403.1456) and [Mantz et al. 2016](https://arxiv.org/abs/1606.03407). Setting $\gamma_\mathrm{scatter} = 0$ recovers the constant-scatter case, and $\gamma_\mathrm{scatter} < 0$ produces **larger scatter at low mass** — physically motivated because low-mass clusters are proportionally more affected by feedback, cool-core disruption, and merger actiation modes.

## Mock cluster catalogue

Draw $N_\mathrm{low}$ + $N_\mathrm{high}$ clusters uniformly in $\log_{10} M$ within each bin, then compute $\ln \mathrm{Obs}$ from the MOR with intrinsic log-normal scatter. Returns arrays in the fit variables $x = \ln(M/M_\mathrm{piv})$ and $y = \ln \mathrm{Obs}$, plus a bin label (0 = low, 1 = high) for bookkeeping.

*Design note.* Uniform in $\log_{10} M$ within each bin is a deliberate simplification: **real halos follow the mass function (tbd)**, so the low edge of each bin is more populated than the high edge. Swapping in HMF-weighted sampling would make the low-mass bin's centroid slightly lower and would amplify the biases below, but doesn't change the qualitative story. Left as an easy extension.

```python
def sigma_of_M(M, sigma0=SIGMA_TRUE, gamma=GAMMA_SCATTER, M_piv=M_PIV):
    """Mass-dependent log-normal scatter.
    sigma_lnO(M) = sigma0 * (M / M_piv) ** gamma
    Following the standard form (Evrard+2014 arXiv:1403.1456; Mantz+2016 arXiv:1606.03407).
    """
    return sigma0 * (M / M_piv) ** gamma


def generate_mock(N_low, N_high, M_edges_low, M_edges_high, rng,
                  gamma=GAMMA_SCATTER):
    logM_low  = rng.uniform(np.log10(M_edges_low[0]),  np.log10(M_edges_low[1]),  N_low)
    logM_high = rng.uniform(np.log10(M_edges_high[0]), np.log10(M_edges_high[1]), N_high)
    logM = np.concatenate([logM_low, logM_high])
    M = 10 ** logM
    x = np.log(M / M_PIV)
    y_mean = A_TRUE + ALPHA_TRUE * x
    # Per-cluster scatter width — this is the only change vs constant-sigma version
    sigma_per_cluster = sigma_of_M(M, sigma0=SIGMA_TRUE, gamma=gamma, M_piv=M_PIV)
    y = y_mean + rng.normal(0.0, 1.0, size=len(x)) * sigma_per_cluster
    bin_label = np.concatenate([np.zeros(N_low, dtype=int),
                                np.ones(N_high, dtype=int)])
    return M, x, y, bin_label
```

## Section 1: Different threshold-based selection cut on observables

### A. One threshold overall (at lower mass bin)

#### 1 — Two fitting procedures

Both operate on the *detected* sample and currently assume constant scatter.

1. **OLS** — closed-form linear regression on $(x_i, y_i)$ with $x = \ln(M/M_\mathrm{piv})$, $y = \ln\mathrm{Obs}$: minimise $\sum_i (y_i - A - \alpha x_i)^2$ for $(A,\alpha)$, then estimate $\sigma$ from the residual RMS ($N-2$ dof, unbiased under Gaussian errors, no truncation). **Assumes a fair draw** — no notion of missing points.

2. **Selection-aware MLE** — maximises the truncated-Gaussian likelihood, which conditions explicitly on detection:
$$p(y \mid x, \text{detected}) = \frac{\mathcal{N}(y \mid A + \alpha x, \sigma)}{P(y > y_\mathrm{min} \mid x)}, \qquad P(y > y_\mathrm{min}\mid x_i) = 1-\Phi\!\left(\frac{y_\mathrm{min}-A-\alpha x_i}{\sigma}\right)$$
As $y_\mathrm{min}\to-\infty$ this reduces to OLS. We minimise the negative log-likelihood
$$-\ln\mathcal{L}(A,\alpha,\sigma) = -\sum_i\left[\ln\phi\!\left(\frac{y_i-A-\alpha x_i}{\sigma}\right) - \ln\sigma - \ln P(y>y_\mathrm{min}\mid x_i)\right]$$
($\phi$ = standard normal PDF) via Nelder–Mead, parametrising as $(A,\alpha,\ln\sigma)$ to keep the sampler unbounded while guaranteeing $\sigma>0$.

We scan $y_\mathrm{min}$ over a grid, run many mock realisations per threshold, and plot fitted $(\sigma,\alpha,A)$ against the resulting total incompleteness $f_\mathrm{total}$.

```python
def fit_ols(x, y):
    """Ordinary least squares; sigma from residual RMS with N-2 dof."""
    if len(x) < 3:
        return np.nan, np.nan, np.nan
    alpha_hat, A_hat = np.polyfit(x, y, 1)   # returns [slope, intercept]
    resid = y - (A_hat + alpha_hat * x)
    sigma_hat = np.std(resid, ddof=2)
    return A_hat, alpha_hat, sigma_hat


def neg_loglike_truncated(params, x, y, y_min):
    """Negative log-likelihood of the truncated-Gaussian MOR."""
    A, alpha, log_sigma = params
    sigma = np.exp(log_sigma)
    mu = A + alpha * x
    log_pdf  = norm.logpdf(y, loc=mu, scale=sigma)
    log_surv = norm.logsf(y_min, loc=mu, scale=sigma)  # ln P(y > y_min | x)
    return -np.sum(log_pdf - log_surv)


def fit_selection_aware(x, y, y_min, init=(0.0, 1.5, 0.2)):
    """MLE of the truncated Gaussian likelihood. `init` = (A, alpha, sigma)."""
    if len(x) < 3:
        return np.nan, np.nan, np.nan
    x0 = [init[0], init[1], np.log(init[2])]
    result = minimize(
        neg_loglike_truncated, x0=x0, args=(x, y, y_min),
        method='Nelder-Mead',
        options={'xatol': 1e-5, 'fatol': 1e-5, 'maxiter': 10000},
    )
    if not result.success:
        return np.nan, np.nan, np.nan
    A, alpha, log_sigma = result.x
    return A, alpha, np.exp(log_sigma)
```

#### 2 — Sanity check: no selection

Both fits should recover truth when $y_\mathrm{min}$ is far below any data. This validates the machinery before we do anything interesting.

```python
# Sanity check FIRST runs with constant scatter (gamma=0) to verify backward compatibility.
M, x, y, bin_lab = generate_mock(1000, 1000, M_EDGES_LOW, M_EDGES_HIGH, rng_master, gamma=0.0)
y_min_none = -1e6

Ao, ao, so = fit_ols(x, y)
As, as_, ss = fit_selection_aware(x, y, y_min_none, init=(A_TRUE, ALPHA_TRUE, SIGMA_TRUE))

print("== Sanity check 1: constant scatter (gamma=0), no selection ==")
print(f"Truth              A={A_TRUE:+.3f}  alpha={ALPHA_TRUE:.3f}  sigma={SIGMA_TRUE:.3f}")
print(f"OLS                A={Ao:+.3f}  alpha={ao:.3f}  sigma={so:.3f}")
print(f"Selection-aware    A={As:+.3f}  alpha={as_:.3f}  sigma={ss:.3f}")

# Now with the actual GAMMA_SCATTER — fits should recover A, alpha, but sigma_fit
# will land at some effective mean between the low-M and high-M scatter values.
M, x, y, bin_lab = generate_mock(1000, 1000, M_EDGES_LOW, M_EDGES_HIGH, rng_master)  # uses GAMMA_SCATTER default
Ao, ao, so = fit_ols(x, y)
As, as_, ss = fit_selection_aware(x, y, y_min_none, init=(A_TRUE, ALPHA_TRUE, SIGMA_TRUE))

print(f"\n== Sanity check 2: mass-dependent scatter (gamma={GAMMA_SCATTER}), no selection ==")
print(f"Truth: A={A_TRUE:+.3f}, alpha={ALPHA_TRUE:.3f}")
print(f"       sigma_lnO(M_piv) = {SIGMA_TRUE:.3f}")
print(f"       sigma_lnO at low bin midpoint  = {sigma_of_M(10**13.85):.3f}")
print(f"       sigma_lnO at high bin midpoint = {sigma_of_M(10**14.85):.3f}")
print(f"OLS                A={Ao:+.3f}  alpha={ao:.3f}  sigma_fit={so:.3f}")
print(f"Selection-aware    A={As:+.3f}  alpha={as_:.3f}  sigma_fit={ss:.3f}")
print("Note: sigma_fit is a single number but truth has different sigma per bin.")
print("      The fit finds some weighted average — this is expected mis-specification.")
```

#### 3 — Plotting

```python
# Reproduce the sketch: one realisation, threshold tuned to ~25-30% total loss.
rng_demo = np.random.default_rng(7)
M, x, y, bin_lab = generate_mock(N_LOW, N_HIGH, M_EDGES_LOW, M_EDGES_HIGH, rng_demo)

Y_MIN_DEMO = -2.1     # roughly f_total ~ 25%
detected = y > Y_MIN_DEMO

Ao, ao, so = fit_ols(x[detected], y[detected])
As, as_, ss = fit_selection_aware(
    x[detected], y[detected], Y_MIN_DEMO,
    init=(A_TRUE, ALPHA_TRUE, SIGMA_TRUE),
)

fig, ax = plt.subplots(figsize=(9, 5.5))
# Swapped axes: ln(Obs) on x, log10(M) on y
ax.scatter(y[detected],  np.log10(M[detected]),  c='black', s=25, label='Detected')
ax.scatter(y[~detected], np.log10(M[~detected]), c='red',   s=25, marker='x', label='Lost')

x_grid = np.linspace(x.min() - 0.05, x.max() + 0.05, 100)
logM_grid = np.log10(M_PIV * np.exp(x_grid))
ax.plot(A_TRUE + ALPHA_TRUE * x_grid, logM_grid,
        color='goldenrod', lw=3, label=fr'Truth: $\alpha$={ALPHA_TRUE:.2f}, $\sigma$={SIGMA_TRUE:.2f}')
ax.plot(Ao + ao * x_grid, logM_grid,
        color='navy', lw=2, label=fr'OLS on detected: $\alpha$={ao:.2f}, $\sigma$={so:.2f}')
ax.plot(As + as_ * x_grid, logM_grid,
        color='green', ls='--', lw=2,
        label=fr'Selection-aware: $\alpha$={as_:.2f}, $\sigma$={ss:.2f}')
# Threshold is now a VERTICAL line at y_min
ax.axvline(Y_MIN_DEMO, color='gray', ls=':', lw=1, label=fr'$y_{{\min}}$ = {Y_MIN_DEMO}')

f_low  = 1 - detected[bin_lab==0].mean()
f_high = 1 - detected[bin_lab==1].mean()
f_tot  = 1 - detected.mean()
ax.set_title(fr'Single mock: $f_\mathrm{{low}}$={f_low:.0%}, $f_\mathrm{{high}}$={f_high:.0%}, $f_\mathrm{{total}}$={f_tot:.0%}')
ax.set_xlabel(r'$\ln\,\mathrm{Obs}$')
ax.set_ylabel(r'$\log_{10}(M / [M_\odot / h])$')
ax.legend(loc='lower right', fontsize=9)
plt.tight_layout()
plt.show()
```

#### 4 — The scan

```python
def scan(y_min_grid, n_realizations, N_low, N_high,
         M_edges_low, M_edges_high,
         seed_offset=0, min_detected=10):
    n_ym = len(y_min_grid)
    keys = ['f_low', 'f_high', 'f_total',
            'A_ols', 'alpha_ols', 'sigma_ols',
            'A_sel', 'alpha_sel', 'sigma_sel']
    results = {k: np.full((n_ym, n_realizations), np.nan) for k in keys}

    for i, y_min in enumerate(y_min_grid):
        for j in range(n_realizations):
            rng = np.random.default_rng(seed_offset + i * 1000 + j)
            M, x, y, bin_lab = generate_mock(N_low, N_high,
                                             M_edges_low, M_edges_high, rng)
            detected = y > y_min

            results['f_low'  ][i, j] = 1 - detected[bin_lab == 0].mean()
            results['f_high' ][i, j] = 1 - detected[bin_lab == 1].mean()
            results['f_total'][i, j] = 1 - detected.mean()

            if detected.sum() < min_detected:
                continue
            xd, yd = x[detected], y[detected]

            Ao, ao, so   = fit_ols(xd, yd)
            As, as_, ss  = fit_selection_aware(
                xd, yd, y_min, init=(A_TRUE, ALPHA_TRUE, SIGMA_TRUE))

            results['A_ols'    ][i, j] = Ao
            results['alpha_ols'][i, j] = ao
            results['sigma_ols'][i, j] = so
            results['A_sel'    ][i, j] = As
            results['alpha_sel'][i, j] = as_
            results['sigma_sel'][i, j] = ss

    return results


Y_MIN_GRID     = np.linspace(-3.5, -1.0, 12)
N_REALIZATIONS = 100

results = scan(Y_MIN_GRID, N_REALIZATIONS,
               N_LOW, N_HIGH,
               M_EDGES_LOW, M_EDGES_HIGH,
               seed_offset=1000)
print(f"Scan complete: {len(Y_MIN_GRID)} thresholds x {N_REALIZATIONS} realisations.")
```

#### 5 — Incompleteness table

It primarily loses the faintest low-mass clusters, and the high-mass end is anchored.

```python
def summary(arr):
    """mean, 16th, 84th percentiles across realisations for each threshold row."""
    return (np.nanmean(arr, axis=1),
            np.nanpercentile(arr, 16, axis=1),
            np.nanpercentile(arr, 84, axis=1))

f_low_m,  _, _ = summary(results['f_low'])
f_high_m, _, _ = summary(results['f_high'])
f_tot_m,  _, _ = summary(results['f_total'])

print(f"{'y_min':>7} {'f_low':>9} {'f_high':>9} {'f_total':>9}")
for ym, fl, fh, ft in zip(Y_MIN_GRID, f_low_m, f_high_m, f_tot_m):
    print(f"{ym:7.2f} {fl:9.1%} {fh:9.1%} {ft:9.1%}")
```

#### 6 — Main result: $\sigma$ vs $f_\mathrm{total}$

```python
s_o_m, s_o_lo, s_o_hi = summary(results['sigma_ols'])
s_s_m, s_s_lo, s_s_hi = summary(results['sigma_sel'])

fig, ax = plt.subplots(figsize=(9, 5.5))
ax.axhline(SIGMA_TRUE, color='goldenrod', ls='--', lw=2,
           label=fr'Truth $\sigma$={SIGMA_TRUE}')

ax.plot(f_tot_m, s_o_m, 'o-', color='navy', lw=2, label='OLS on detected')
ax.fill_between(f_tot_m, s_o_lo, s_o_hi, color='navy', alpha=0.2)

ax.plot(f_tot_m, s_s_m, 's-', color='green', lw=2, label='Selection-aware MLE')
ax.fill_between(f_tot_m, s_s_lo, s_s_hi, color='green', alpha=0.2)

ax.set_xlabel(r'$f_\mathrm{total}$ (fraction of clusters lost)')
ax.set_ylabel(r'Fitted $\sigma_{\ln\,\mathrm{Obs}}$')
ax.set_title('Recovered intrinsic scatter vs total incompleteness')
ax.legend(loc='best')
plt.tight_layout()
plt.show()
```

#### 7 — Companion panels: $\alpha$ and $A$ vs $f_\mathrm{total}$

```python
a_o_m, a_o_lo, a_o_hi = summary(results['alpha_ols'])
a_s_m, a_s_lo, a_s_hi = summary(results['alpha_sel'])
A_o_m, A_o_lo, A_o_hi = summary(results['A_ols'])
A_s_m, A_s_lo, A_s_hi = summary(results['A_sel'])

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.8))

ax1.axhline(ALPHA_TRUE, color='goldenrod', ls='--', lw=2, label=fr'Truth $\alpha$={ALPHA_TRUE}')
ax1.plot(f_tot_m, a_o_m, 'o-', color='navy', lw=2, label='OLS')
ax1.fill_between(f_tot_m, a_o_lo, a_o_hi, color='navy', alpha=0.2)
ax1.plot(f_tot_m, a_s_m, 's-', color='green', lw=2, label='Selection-aware')
ax1.fill_between(f_tot_m, a_s_lo, a_s_hi, color='green', alpha=0.2)
ax1.set_xlabel(r'$f_\mathrm{total}$')
ax1.set_ylabel(r'Fitted slope $\alpha$')
ax1.set_title('Slope')
ax1.legend()

ax2.axhline(A_TRUE, color='goldenrod', ls='--', lw=2, label=fr'Truth $A$={A_TRUE}')
ax2.plot(f_tot_m, A_o_m, 'o-', color='navy', lw=2, label='OLS')
ax2.fill_between(f_tot_m, A_o_lo, A_o_hi, color='navy', alpha=0.2)
ax2.plot(f_tot_m, A_s_m, 's-', color='green', lw=2, label='Selection-aware')
ax2.fill_between(f_tot_m, A_s_lo, A_s_hi, color='green', alpha=0.2)
ax2.set_xlabel(r'$f_\mathrm{total}$')
ax2.set_ylabel(r'Fitted intercept $A$')
ax2.set_title('Intercept')
ax2.legend()

plt.tight_layout()
plt.show()
```

## B. bin-wise fraction loss

e.

**Default ratio.** From our earlier discussion: $f_\mathrm{low} : f_\mathrm{high} = 2 : 1$, so at each scan point $f_\mathrm{high} = f_\mathrm{low}/2$. This encodes the expectation that lower-mass, fainter clusters are always harder to detect. We scan $f_\mathrm{low} \in [0, 0.5]$, giving $f_\mathrm{total} \in [0, 37.5\%]$ (weighted average with equal $N$ per bin).

**Selection-aware fit under fraction selection.** Fraction selection is equivalent to a *bin-dependent* threshold: within bin $k$, the effective $y_\mathrm{min}^{(k)}$ is the observable value of the lowest-$y$ surviving cluster in that bin. So the truncated-Gaussian likelihood generalises to
$$-\ln \mathcal{L}(A, \alpha, \sigma) = -\sum_i \left[ \ln \phi\!\left(\tfrac{y_i - A - \alpha x_i}{\sigma}\right) - \ln \sigma - \ln P(y > y_\mathrm{min}^{(k_i)} \mid x_i) \right]$$

where $k_i$ is the bin of point $i$. We just pass a *per-point* $y_\mathrm{min}$ array to `norm.logsf`.

*Caveat.* Using the empirical $y_\mathrm{min}$ from the same mock as the fit is slightly circular — in a real analysis you'd measure completeness from external injection tests, not from the sample itself. For the pedagogy here it's fine; the answer converges to the "true" bin-dependent threshold as $N$ grows.


```python
def fraction_based_selection(y, bin_lab, f_low, f_high):
    """Remove the lowest-y f_low fraction from bin 0, f_high from bin 1.

    Returns:
        detected : bool array of survivors
        (y_min_low, y_min_high) : empirical thresholds per bin (=y of lowest survivor)
    """
    detected = np.ones(len(y), dtype=bool)
    y_min_per_bin = [-np.inf, -np.inf]

    for k, f in [(0, f_low), (1, f_high)]:
        idx = np.where(bin_lab == k)[0]
        if len(idx) == 0 or f <= 0:
            continue
        n_remove = int(round(f * len(idx)))
        if n_remove == 0:
            continue
        if n_remove >= len(idx):
            detected[idx] = False
            continue
        order = np.argsort(y[idx])
        detected[idx[order[:n_remove]]] = False
        y_min_per_bin[k] = y[idx[order[n_remove:]]].min()

    return detected, tuple(y_min_per_bin)


def neg_loglike_truncated_perpoint(params, x, y, y_min_arr):
    """Truncated-Gaussian negative log-likelihood with a per-point y_min."""
    A, alpha, log_sigma = params
    sigma = np.exp(log_sigma)
    mu = A + alpha * x
    log_pdf  = norm.logpdf(y, loc=mu, scale=sigma)
    log_surv = norm.logsf(y_min_arr, loc=mu, scale=sigma)
    return -np.sum(log_pdf - log_surv)


def fit_selection_aware_perpoint(x, y, y_min_arr, init=(0.0, 1.5, 0.2)):
    if len(x) < 3:
        return np.nan, np.nan, np.nan
    x0 = [init[0], init[1], np.log(init[2])]
    r = minimize(
        neg_loglike_truncated_perpoint, x0=x0, args=(x, y, y_min_arr),
        method='Nelder-Mead',
        options={'xatol': 1e-5, 'fatol': 1e-5, 'maxiter': 10000},
    )
    if not r.success:
        return np.nan, np.nan, np.nan
    A, alpha, log_sigma = r.x
    return A, alpha, np.exp(log_sigma)
```

### Diagnostic: single mock under fraction-based selection

One realisation at $(f_\mathrm{low}, f_\mathrm{high}) = (30\%, 15\%)$r.

### Fraction-based scan

. Here the scan variable is $f_\mathrm{low}$ with $f_\mathrm{high} = f_\mathrm{low} / 2$.

```python
def scan_fraction(f_low_grid, ratio, n_realizations, N_low, N_high,
                  M_edges_low, M_edges_high,
                  seed_offset=2000, min_detected=10):
    """Scan fraction-based selection. f_high = ratio * f_low at each step."""
    n = len(f_low_grid)
    keys = ['f_low', 'f_high', 'f_total',
            'A_ols', 'alpha_ols', 'sigma_ols',
            'A_sel', 'alpha_sel', 'sigma_sel']
    results = {k: np.full((n, n_realizations), np.nan) for k in keys}

    for i, fl in enumerate(f_low_grid):
        fh = ratio * fl
        for j in range(n_realizations):
            rng = np.random.default_rng(seed_offset + i * 1000 + j)
            M, x, y, bin_lab = generate_mock(N_low, N_high,
                                             M_edges_low, M_edges_high, rng)
            det, (ym_lo, ym_hi) = fraction_based_selection(y, bin_lab, fl, fh)

            results['f_low'  ][i, j] = 1 - det[bin_lab == 0].mean()
            results['f_high' ][i, j] = 1 - det[bin_lab == 1].mean()
            results['f_total'][i, j] = 1 - det.mean()

            if det.sum() < min_detected:
                continue

            xd, yd = x[det], y[det]
            Ao, ao, so = fit_ols(xd, yd)

            y_min_arr = np.where(bin_lab[det] == 0, ym_lo, ym_hi)
            As, as_, ss = fit_selection_aware_perpoint(
                xd, yd, y_min_arr,
                init=(A_TRUE, ALPHA_TRUE, SIGMA_TRUE),
            )

            results['A_ols'    ][i, j] = Ao
            results['alpha_ols'][i, j] = ao
            results['sigma_ols'][i, j] = so
            results['A_sel'    ][i, j] = As
            results['alpha_sel'][i, j] = as_
            results['sigma_sel'][i, j] = ss

    return results


F_LOW_GRID = np.linspace(0.0, 0.5, 11)
RATIO_HIGH_OVER_LOW = 0.5   # f_high = 0.5 * f_low
results_frac = scan_fraction(
    F_LOW_GRID, RATIO_HIGH_OVER_LOW, N_REALIZATIONS,
    N_LOW, N_HIGH, M_EDGES_LOW, M_EDGES_HIGH,
    seed_offset=2000,
)
print(f"Fraction scan complete: {len(F_LOW_GRID)} f_low values x {N_REALIZATIONS} realisations.")

# Table
f_low_fm, _, _  = summary(results_frac['f_low'])
f_high_fm, _, _ = summary(results_frac['f_high'])
f_tot_fm, _, _  = summary(results_frac['f_total'])
print(f"\n{'f_low set':>10} {'f_low real':>11} {'f_high real':>12} {'f_total':>9}")
for fl_set, fl, fh, ft in zip(F_LOW_GRID, f_low_fm, f_high_fm, f_tot_fm):
    print(f"{fl_set:10.2f} {fl:11.1%} {fh:12.1%} {ft:9.1%}")
```

## C. Null test: observable-independent random lossthe fit.

```python
def random_selection(y, bin_lab, f_low, f_high, rng):
    """Randomly remove f_low fraction from bin 0, f_high from bin 1.
    Selection is INDEPENDENT of the observable — this is the null test.
    """
    detected = np.ones(len(y), dtype=bool)
    for k, f in [(0, f_low), (1, f_high)]:
        idx = np.where(bin_lab == k)[0]
        if len(idx) == 0 or f <= 0:
            continue
        n_remove = int(round(f * len(idx)))
        if n_remove == 0:
            continue
        if n_remove >= len(idx):
            detected[idx] = False
            continue
        to_remove = rng.choice(idx, size=n_remove, replace=False)
        detected[to_remove] = False
    return detected
```

## Section 2:  Mass dependent scatter and slope(tbd)

The scatter is now mass-dependent by construction (γ = −0.3). Two hypotheses about what a real analyst would do:

**H1 — "we don't know it's mass-dependent."** Fit assuming constant σ. Three free parameters: (A, α, σ). The model is *structurally mis-specified*: no single σ can describe truth that has different scatter in the two bins.

**H2 — "we know it's a power law in mass, but we don't know σ_0 or γ."** Fit with σ_i = σ_0 · (M_i/M_piv)^γ. Four free parameters: (A, α, σ_0, γ). The model is *correctly specified*.

Each hypothesis gets two selection treatments: OLS-style (no truncation modelled) and selection-aware (truncation modelled). That gives four fits per mock:

| Fit | Scatter model | Selection model |
|---|---|---|
| **H1-OLS** | constant σ | none — assumes fair sample |
| **H1-Sel** | constant σ | truncated Gaussian |
| **H2-MLE** | mass-dep σ | none |
| **H2-Sel** | mass-dep σ | trunc (γ_model floated).

```python
GAMMA_SCATTER = -0.3    # try 0.0, -0.3, -0.6 to see the effect grow
# ---------------------------------------------------------------------------
# H2 model: mass-dependent scatter, sigma_i = sigma_0 * (M_i/M_piv)^gamma
# Since x = ln(M/M_piv), this is equivalently sigma_i = sigma_0 * exp(gamma * x_i).
# H2 fits estimate 4 parameters: (A, alpha, sigma_0, gamma).
# ---------------------------------------------------------------------------

def neg_loglike_h2_untrunc(params, x, y):
    """Un-truncated Gaussian log-likelihood with mass-dependent sigma.
    Parameterisation: log_sigma0 (unbounded internal), gamma (unbounded).
    """
    A, alpha, log_sigma0, gamma = params
    sigma0 = np.exp(log_sigma0)
    sigma_i = sigma0 * np.exp(gamma * x)   # per-cluster width
    mu = A + alpha * x
    return -np.sum(norm.logpdf(y, loc=mu, scale=sigma_i))


def fit_h2_untrunc(x, y, init=(0.0, 1.5, 0.2, 0.0)):
    """H2 fit WITHOUT selection modeling. Heteroscedastic Gaussian MLE.
    Analogue of OLS for the mass-dependent-scatter model.
    init: (A_init, alpha_init, sigma0_init, gamma_init).
    """
    if len(x) < 4:
        return np.nan, np.nan, np.nan, np.nan
    x0 = [init[0], init[1], np.log(init[2]), init[3]]
    r = minimize(neg_loglike_h2_untrunc, x0=x0, args=(x, y),
                 method='Nelder-Mead',
                 options={'xatol': 1e-5, 'fatol': 1e-5, 'maxiter': 20000})
    if not r.success:
        return np.nan, np.nan, np.nan, np.nan
    A, alpha, log_sigma0, gamma = r.x
    return A, alpha, np.exp(log_sigma0), gamma


def neg_loglike_h2_truncated(params, x, y, y_min):
    """Truncated-Gaussian log-likelihood with mass-dependent sigma.
    Same structure as H1 selection-aware, but sigma varies per point.
    """
    A, alpha, log_sigma0, gamma = params
    sigma0 = np.exp(log_sigma0)
    sigma_i = sigma0 * np.exp(gamma * x)
    mu = A + alpha * x
    log_pdf  = norm.logpdf(y, loc=mu, scale=sigma_i)
    log_surv = norm.logsf(y_min, loc=mu, scale=sigma_i)   # per-point survival prob
    return -np.sum(log_pdf - log_surv)


def fit_h2_sel(x, y, y_min, init=(0.0, 1.5, 0.2, 0.0)):
    """H2 fit WITH selection modeling. Correctly specified when both mass-dep
    scatter is present AND selection is threshold-based.
    """
    if len(x) < 4:
        return np.nan, np.nan, np.nan, np.nan
    x0 = [init[0], init[1], np.log(init[2]), init[3]]
    r = minimize(neg_loglike_h2_truncated, x0=x0, args=(x, y, y_min),
                 method='Nelder-Mead',
                 options={'xatol': 1e-5, 'fatol': 1e-5, 'maxiter': 20000})
    if not r.success:
        return np.nan, np.nan, np.nan, np.nan
    A, alpha, log_sigma0, gamma = r.x
    return A, alpha, np.exp(log_sigma0), gamma

def neg_loglike_h2_truncated_perpoint(params, x, y, y_min_arr):
    """H2 truncated-Gaussian log-likelihood with per-point y_min AND per-point sigma.
    Combines the per-point threshold structure (needed for fraction-based selection)
    with mass-dependent scatter.
    """
    A, alpha, log_sigma0, gamma = params
    sigma0 = np.exp(log_sigma0)
    sigma_i = sigma0 * np.exp(gamma * x)
    mu = A + alpha * x
    log_pdf  = norm.logpdf(y, loc=mu, scale=sigma_i)
    log_surv = norm.logsf(y_min_arr, loc=mu, scale=sigma_i)
    return -np.sum(log_pdf - log_surv)


def fit_h2_sel_perpoint(x, y, y_min_arr, init=(0.0, 1.5, 0.2, 0.0)):
    """H2 selection-aware fit with per-point y_min. For fraction-based selection."""
    if len(x) < 4:
        return np.nan, np.nan, np.nan, np.nan
    x0 = [init[0], init[1], np.log(init[2]), init[3]]
    r = minimize(neg_loglike_h2_truncated_perpoint, x0=x0, args=(x, y, y_min_arr),
                 method='Nelder-Mead',
                 options={'xatol': 1e-5, 'fatol': 1e-5, 'maxiter': 20000})
    if not r.success:
        return np.nan, np.nan, np.nan, np.nan
    A, alpha, log_sigma0, gamma = r.x
    return A, alpha, np.exp(log_sigma0), gamma


def hybrid_selection(y, bin_lab, y_min_low, f_high, rng):
    """Hybrid selection: threshold cut in low bin, random loss in high bin.

    Low bin (bin_lab == 0): keep only clusters with y > y_min_low  (Malmquist-type).
    High bin (bin_lab == 1): keep a random (1 - f_high) fraction    (observable-independent).

    Rationale: low-mass clusters are near the survey flux limit, so their loss
    is observable-dependent. High-mass clusters are always above the flux limit,
    so any loss there (misidentification, follow-up failure) is uncorrelated
    with the observable.
    """
    detected = np.ones(len(y), dtype=bool)

    # Low bin: observable threshold
    low_mask = (bin_lab == 0)
    detected[low_mask & (y <= y_min_low)] = False

    # High bin: random loss
    high_idx = np.where(bin_lab == 1)[0]
    if len(high_idx) > 0 and f_high > 0:
        n_remove = int(round(f_high * len(high_idx)))
        if n_remove > 0:
            to_remove = rng.choice(high_idx, size=n_remove, replace=False)
            detected[to_remove] = False

    return detected
```

### Different selection for each bin

```python
def scan_h1h2_hybrid(y_min_grid, f_high, n_realizations, N_low, N_high,
                     M_edges_low, M_edges_high, gamma_truth=GAMMA_SCATTER,
                     seed_offset=11000, min_detected=10):
    """H1/H2 scan under hybrid selection: threshold low-bin, random loss high-bin.

    Scan variable: y_min_low  (the low-bin observable cut).
    Held fixed:    f_high     (random loss fraction in high bin).

    The selection-aware fits use y_min_low as their assumed threshold.
    Because high-bin y values are far above any reasonable y_min_low, the
    threshold effectively affects only the low bin — the fit's likelihood
    handles the high bin as untruncated, which is correct (random loss is
    observable-independent so it drops out of the MLE).
    """
    n = len(y_min_grid)
    keys = ['f_low', 'f_high', 'f_total',
            'A_h1o', 'a_h1o', 's_h1o',
            'A_h1s', 'a_h1s', 's_h1s',
            'A_h2m', 'a_h2m', 's_h2m', 'g_h2m',
            'A_h2s', 'a_h2s', 's_h2s', 'g_h2s']
    results = {k: np.full((n, n_realizations), np.nan) for k in keys}

    init_h1 = (A_TRUE, ALPHA_TRUE, SIGMA_TRUE)
    init_h2 = (A_TRUE, ALPHA_TRUE, SIGMA_TRUE, gamma_truth)

    for i, y_min in enumerate(y_min_grid):
        for j in range(n_realizations):
            rng = np.random.default_rng(seed_offset + i * 1000 + j)
            M, x, y, bin_lab = generate_mock(N_low, N_high,
                                             M_edges_low, M_edges_high, rng,
                                             gamma=gamma_truth)
            det = hybrid_selection(y, bin_lab, y_min, f_high, rng)

            results['f_low'  ][i, j] = 1 - det[bin_lab == 0].mean()
            results['f_high' ][i, j] = 1 - det[bin_lab == 1].mean()
            results['f_total'][i, j] = 1 - det.mean()
            if det.sum() < min_detected:
                continue
            xd, yd = x[det], y[det]

            Ao,   ao,   so                = fit_ols(xd, yd)
            Ah1,  ah1,  sh1               = fit_selection_aware(xd, yd, y_min, init=init_h1)
            Ah2m, ah2m, s0h2m, gh2m       = fit_h2_untrunc(xd, yd, init=init_h2)
            Ah2s, ah2s, s0h2s, gh2s       = fit_h2_sel(xd, yd, y_min, init=init_h2)

            results['A_h1o'][i,j], results['a_h1o'][i,j], results['s_h1o'][i,j] = Ao, ao, so
            results['A_h1s'][i,j], results['a_h1s'][i,j], results['s_h1s'][i,j] = Ah1, ah1, sh1
            results['A_h2m'][i,j], results['a_h2m'][i,j], results['s_h2m'][i,j], results['g_h2m'][i,j] = Ah2m, ah2m, s0h2m, gh2m
            results['A_h2s'][i,j], results['a_h2s'][i,j], results['s_h2s'][i,j], results['g_h2s'][i,j] = Ah2s, ah2s, s0h2s, gh2s
    return results


# Run it — same y_min grid as your threshold scan, moderate 15% random loss in high bin
F_HIGH_HYBRID = 0.15
results_h1h2_hybrid = scan_h1h2_hybrid(
    Y_MIN_GRID, F_HIGH_HYBRID, N_REALIZATIONS,
    N_LOW, N_HIGH, M_EDGES_LOW, M_EDGES_HIGH,
    gamma_truth=GAMMA_SCATTER, seed_offset=11000,
)
print(f"Hybrid scan complete: threshold low-bin + random {F_HIGH_HYBRID:.0%} high-bin.")
```

### Diagnostic: fitted MOR

One realisation at threshold $y_\mathrm{min} = -2.1$ ($f_\mathrm{total} \approx 30\%$). Both panels show the same data and same truth. The two panels differ only in *which family of fits* is drawn on top:

- **Left panel (H1)**: constant-$\sigma$ fits. Their $\pm 1\sigma$ envelopes are **parallel** — same width at every mass by construction.
- **Right panel (H2)**: mass-dependent-$\sigma$ fits. Their envelopes are **wedge-shaped**, matching the truth envelope's structure.

The truth envelope (goldenrod, both panels) is wedge-shaped because $\gamma = -0.3$ makes $\sigma$ larger at low mass. The visual mismatch between H1's parallel envelopes and the truth's wedge is the *pictorial signature* of scatter-model mis-specification. H2-Sel recovers not just the mean line but the whole envelope shape.

```python
FITS_STYLE = [
    ('H1-OLS',  'a_h1o', 's_h1o', 'A_h1o', None,     'navy',       'o', '-',  0.9),
    ('H1-Sel',  'a_h1s', 's_h1s', 'A_h1s', None,     'green',      's', '-',  0.9),
    ('H2-MLE',  'a_h2m', 's_h2m', 'A_h2m', 'g_h2m',  'crimson',    '^', '--', 0.9),
    ('H2-Sel',  'a_h2s', 's_h2s', 'A_h2s', 'g_h2s',  'darkorange', 'D', '--', 0.9),
]

def summary(a):
    return (np.nanmean(a, axis=1),
            np.nanpercentile(a, 16, axis=1),
            np.nanpercentile(a, 84, axis=1))


def plot_mean_relation_scan(results, title_suffix=''):
    """Two-panel scan plot: alpha and A vs f_total, four fits each."""
    f_tot_m, _, _ = summary(results['f_total'])
    fig, (ax_a, ax_A) = plt.subplots(1, 2, figsize=(13, 4.8))

    for label, akey, _, Akey, _, color, mk, ls, alpha in FITS_STYLE:
        a_m, a_lo, a_hi = summary(results[akey])
        A_m, A_lo, A_hi = summary(results[Akey])
        ax_a.plot(f_tot_m, a_m, marker=mk, ls=ls, color=color, lw=2, alpha=alpha, label=label)
        ax_a.fill_between(f_tot_m, a_lo, a_hi, color=color, alpha=0.12)
        ax_A.plot(f_tot_m, A_m, marker=mk, ls=ls, color=color, lw=2, alpha=alpha, label=label)
        ax_A.fill_between(f_tot_m, A_lo, A_hi, color=color, alpha=0.12)

    ax_a.axhline(ALPHA_TRUE, color='goldenrod', ls=':', lw=2, label=fr'Truth $\alpha$={ALPHA_TRUE}')
    ax_a.set_xlabel(r'$f_\mathrm{total}$'); ax_a.set_ylabel(r'Fitted slope $\alpha$')
    ax_a.set_title(f'Slope{title_suffix}'); ax_a.legend(fontsize=9)

    ax_A.axhline(A_TRUE, color='goldenrod', ls=':', lw=2, label=fr'Truth $A$={A_TRUE}')
    ax_A.set_xlabel(r'$f_\mathrm{total}$'); ax_A.set_ylabel(r'Fitted intercept $A$')
    ax_A.set_title(f'Intercept{title_suffix}'); ax_A.legend(fontsize=9)
    plt.tight_layout(); plt.show()


def plot_scatter_scan(results, title_suffix=''):
    """Three-panel scan plot: H1 sigma, H2 sigma_0, H2 gamma."""
    f_tot_m, _, _ = summary(results['f_total'])
    M_low_mid  = np.sqrt(M_EDGES_LOW[0]  * M_EDGES_LOW[1])
    M_high_mid = np.sqrt(M_EDGES_HIGH[0] * M_EDGES_HIGH[1])
    sig_tr_low  = sigma_of_M(M_low_mid,  sigma0=SIGMA_TRUE, gamma=GAMMA_SCATTER)
    sig_tr_high = sigma_of_M(M_high_mid, sigma0=SIGMA_TRUE, gamma=GAMMA_SCATTER)

    fig, (ax_s1, ax_s0, ax_g) = plt.subplots(1, 3, figsize=(16, 4.8))

    for label, _, skey, _, _, color, mk, ls, alpha in FITS_STYLE[:2]:
        s_m, s_lo, s_hi = summary(results[skey])
        ax_s1.plot(f_tot_m, s_m, marker=mk, ls=ls, color=color, lw=2, alpha=alpha, label=label)
        ax_s1.fill_between(f_tot_m, s_lo, s_hi, color=color, alpha=0.15)
    ax_s1.axhline(sig_tr_low,  color='goldenrod', ls=':',  lw=1.7,
                  label=fr'$\sigma_\mathrm{{true}}$(low)={sig_tr_low:.2f}')
    ax_s1.axhline(sig_tr_high, color='goldenrod', ls='-.', lw=1.7,
                  label=fr'$\sigma_\mathrm{{true}}$(high)={sig_tr_high:.2f}')
    ax_s1.set_xlabel(r'$f_\mathrm{total}$'); ax_s1.set_ylabel(r'Fitted $\sigma$ (const-$\sigma$)')
    ax_s1.set_title(f'H1: $\\sigma$ (mis-specified){title_suffix}'); ax_s1.legend(fontsize=8)

    for label, _, skey, _, _, color, mk, ls, alpha in FITS_STYLE[2:]:
        s_m, s_lo, s_hi = summary(results[skey])
        ax_s0.plot(f_tot_m, s_m, marker=mk, ls=ls, color=color, lw=2, alpha=alpha, label=label)
        ax_s0.fill_between(f_tot_m, s_lo, s_hi, color=color, alpha=0.15)
    ax_s0.axhline(SIGMA_TRUE, color='goldenrod', ls=':', lw=2,
                  label=fr'Truth $\sigma_0$={SIGMA_TRUE}')
    ax_s0.set_xlabel(r'$f_\mathrm{total}$'); ax_s0.set_ylabel(r'Fitted $\sigma_0$')
    ax_s0.set_title(f'H2: $\\sigma_0${title_suffix}'); ax_s0.legend(fontsize=8)

    for label, _, _, _, gkey, color, mk, ls, alpha in FITS_STYLE[2:]:
        g_m, g_lo, g_hi = summary(results[gkey])
        ax_g.plot(f_tot_m, g_m, marker=mk, ls=ls, color=color, lw=2, alpha=alpha, label=label)
        ax_g.fill_between(f_tot_m, g_lo, g_hi, color=color, alpha=0.15)
    ax_g.axhline(GAMMA_SCATTER, color='goldenrod', ls=':', lw=2,
                 label=fr'Truth $\gamma$={GAMMA_SCATTER}')
    ax_g.axhline(0.0, color='gray', ls='-', lw=0.7, alpha=0.4)
    ax_g.set_xlabel(r'$f_\mathrm{total}$'); ax_g.set_ylabel(r'Fitted $\gamma$')
    ax_g.set_title(f'H2: $\\gamma${title_suffix}'); ax_g.legend(fontsize=8)
    plt.tight_layout(); plt.show()


def plot_diagnostic_mor(M, x, y, bin_lab, detected, fits, selection_label,
                        threshold_marker=None, thresholds_per_bin=None):
    """Two-panel diagnostic: fitted MOR lines and envelopes on top of the mock.

    fits: dict with keys 'h1_ols', 'h1_sel', 'h2_mle', 'h2_sel', each a tuple of parameters:
          - H1 fits: (A, alpha, sigma)
          - H2 fits: (A, alpha, sigma0, gamma)
    selection_label: title suffix, e.g. 'threshold', 'fraction', 'random'.
    threshold_marker: single y_min value to plot as a vertical line (for threshold-based).
    thresholds_per_bin: (y_min_low, y_min_high) for fraction-based, or None.
    """
    logM_grid = np.linspace(np.log10(M_EDGES_LOW[0]) - 0.05,
                            np.log10(M_EDGES_HIGH[1]) + 0.05, 200)
    x_grid = np.log(10**logM_grid / M_PIV)

    def _line(A, a, xg): return A + a * xg
    def _env_const(A, a, s, xg):
        m = _line(A, a, xg); return m - s, m + s
    def _env_massdep(A, a, s0, g, xg):
        m = _line(A, a, xg); s_i = s0 * np.exp(g * xg); return m - s_i, m + s_i

    fig, (axL, axR) = plt.subplots(1, 2, figsize=(15, 6), sharey=True)
    for ax in (axL, axR):
        ax.scatter(y[detected],  np.log10(M[detected]),  c='black', s=22, label='Detected')
        ax.scatter(y[~detected], np.log10(M[~detected]), c='red',   s=22, marker='x', label='Lost')
        if threshold_marker is not None:
            ax.axvline(threshold_marker, color='gray', ls=':', lw=1,
                       label=fr'$y_{{\min}}$={threshold_marker}')
        if thresholds_per_bin is not None:
            ym_lo, ym_hi = thresholds_per_bin
            ax.axvline(ym_lo, color='gray', ls=':', lw=1, alpha=0.7,
                       label=fr'$y_{{\min}}^\mathrm{{low}}$={ym_lo:.2f}')
            ax.axvline(ym_hi, color='gray', ls='-.', lw=1, alpha=0.7,
                       label=fr'$y_{{\min}}^\mathrm{{high}}$={ym_hi:.2f}')
        # Truth
        mean_tr = _line(A_TRUE, ALPHA_TRUE, x_grid)
        lo_tr, hi_tr = _env_massdep(A_TRUE, ALPHA_TRUE, SIGMA_TRUE, GAMMA_SCATTER, x_grid)
        ax.plot(mean_tr, logM_grid, color='goldenrod', lw=2.5,
                label=fr'Truth ($\alpha$={ALPHA_TRUE}, $\gamma$={GAMMA_SCATTER})')
        ax.fill_betweenx(logM_grid, lo_tr, hi_tr, color='goldenrod', alpha=0.18,
                         label=r'Truth $\pm1\sigma$')
        ax.set_xlabel(r'$\ln\,\mathrm{Obs}$')

    # H1 panel
    for key, color, ls, name in [('h1_ols', 'navy', '-', 'H1-OLS'),
                                  ('h1_sel', 'green', '--', 'H1-Sel')]:
        A, a, s = fits[key]
        lo, hi = _env_const(A, a, s, x_grid)
        axL.plot(_line(A, a, x_grid), logM_grid, color=color, ls=ls, lw=2,
                 label=fr'{name} ($\alpha$={a:.2f}, $\sigma$={s:.2f})')
        axL.fill_betweenx(logM_grid, lo, hi, color=color, alpha=0.10)

    # H2 panel
    for key, color, ls, name in [('h2_mle',  'crimson',    '-',  'H2-MLE'),
                                  ('h2_sel',  'darkorange', '--', 'H2-Sel')]:
        A, a, s0, g = fits[key]
        lo, hi = _env_massdep(A, a, s0, g, x_grid)
        axR.plot(_line(A, a, x_grid), logM_grid, color=color, ls=ls, lw=2,
                 label=fr'{name} ($\alpha$={a:.2f}, $\sigma_0$={s0:.2f}, $\gamma$={g:+.2f})')
        axR.fill_betweenx(logM_grid, lo, hi, color=color, alpha=0.10)

    axL.set_title(r'H1: constant-$\sigma$ fits')
    axR.set_title(r'H2: mass-dependent-$\sigma$ fits')
    axL.set_ylabel(r'$\log_{10}(M / [M_\odot / h])$')
    for ax in (axL, axR):
        ax.legend(loc='upper left', fontsize=8)

    f_low_d  = 1 - detected[bin_lab==0].mean()
    f_high_d = 1 - detected[bin_lab==1].mean()
    f_tot_d  = 1 - detected.mean()
    plt.suptitle(fr'{selection_label}: $f_\mathrm{{low}}$={f_low_d:.0%}, '
                 fr'$f_\mathrm{{high}}$={f_high_d:.0%}, $f_\mathrm{{total}}$={f_tot_d:.0%}',
                 fontsize=11)
    plt.tight_layout(); plt.show()
```

```python
rng_diag = np.random.default_rng(14)
M, x, y, bin_lab = generate_mock(N_LOW, N_HIGH, M_EDGES_LOW, M_EDGES_HIGH, rng_diag,
                                  gamma=GAMMA_SCATTER)

Y_MIN_HYBRID_DEMO   = -2.1     # low-bin threshold
F_HIGH_HYBRID_DEMO  = 0.15     # high-bin random loss fraction
det = hybrid_selection(y, bin_lab, Y_MIN_HYBRID_DEMO, F_HIGH_HYBRID_DEMO, rng_diag)

init_h1 = (A_TRUE, ALPHA_TRUE, SIGMA_TRUE)
init_h2 = (A_TRUE, ALPHA_TRUE, SIGMA_TRUE, GAMMA_SCATTER)

Ao,   ao,   so                = fit_ols(x[det], y[det])
Ah1,  ah1,  sh1               = fit_selection_aware(x[det], y[det], Y_MIN_HYBRID_DEMO, init=init_h1)
Ah2m, ah2m, s0h2m, gh2m       = fit_h2_untrunc(x[det], y[det], init=init_h2)
Ah2s, ah2s, s0h2s, gh2s       = fit_h2_sel(x[det], y[det], Y_MIN_HYBRID_DEMO, init=init_h2)

# Colour-code the lost points by loss mechanism, so the two selection modes are visually distinct
lost_low  = (~det) & (bin_lab == 0)   # dropped by threshold (Malmquist)
lost_high = (~det) & (bin_lab == 1)   # dropped by random loss

logM_grid = np.linspace(np.log10(M_EDGES_LOW[0]) - 0.05,
                        np.log10(M_EDGES_HIGH[1]) + 0.05, 200)
x_grid = np.log(10**logM_grid / M_PIV)

def _line(A, a, xg):  return A + a * xg
def _env_const(A, a, s, xg):
    m = _line(A, a, xg);  return m - s, m + s
def _env_massdep(A, a, s0, g, xg):
    m = _line(A, a, xg);  s_i = s0 * np.exp(g * xg);  return m - s_i, m + s_i

fig, (axL, axR) = plt.subplots(1, 2, figsize=(15, 6), sharey=True)

for ax in (axL, axR):
    ax.scatter(y[det], np.log10(M[det]), c='black', s=22, label='Detected')
    ax.scatter(y[lost_low],  np.log10(M[lost_low]),
               c='red', s=25, marker='x',
               label='Lost — low bin (threshold)')
    ax.scatter(y[lost_high], np.log10(M[lost_high]),
               s=45, marker='o', facecolors='none',
               edgecolors='steelblue', linewidths=1.5,
               label='Lost — high bin (random)')
    ax.axvline(Y_MIN_HYBRID_DEMO, color='gray', ls=':', lw=1,
               label=fr'$y_{{\min}}^\mathrm{{low}}$={Y_MIN_HYBRID_DEMO}')

    # Truth mean + envelope
    mean_tr = _line(A_TRUE, ALPHA_TRUE, x_grid)
    lo_tr, hi_tr = _env_massdep(A_TRUE, ALPHA_TRUE, SIGMA_TRUE, GAMMA_SCATTER, x_grid)
    ax.plot(mean_tr, logM_grid, color='goldenrod', lw=2.5,
            label=fr'Truth ($\alpha$={ALPHA_TRUE}, $\gamma$={GAMMA_SCATTER})')
    ax.fill_betweenx(logM_grid, lo_tr, hi_tr, color='goldenrod', alpha=0.18,
                     label=r'Truth $\pm 1\sigma$')
    ax.set_xlabel(r'$\ln\,\mathrm{Obs}$')

# Left panel — H1 fits
for A, a, s, color, ls, name in [
    (Ao,  ao,  so,  'navy',  '-',  'H1-OLS'),
    (Ah1, ah1, sh1, 'green', '--', 'H1-Sel'),
]:
    lo, hi = _env_const(A, a, s, x_grid)
    axL.plot(_line(A, a, x_grid), logM_grid, color=color, ls=ls, lw=2,
             label=fr'{name} ($\alpha$={a:.2f}, $\sigma$={s:.2f})')
    axL.fill_betweenx(logM_grid, lo, hi, color=color, alpha=0.10)

# Right panel — H2 fits
for A, a, s0, g, color, ls, name in [
    (Ah2m, ah2m, s0h2m, gh2m, 'crimson',    '-',  'H2-MLE'),
    (Ah2s, ah2s, s0h2s, gh2s, 'darkorange', '--', 'H2-Sel'),
]:
    lo, hi = _env_massdep(A, a, s0, g, x_grid)
    axR.plot(_line(A, a, x_grid), logM_grid, color=color, ls=ls, lw=2,
             label=fr'{name} ($\alpha$={a:.2f}, $\sigma_0$={s0:.2f}, $\gamma$={g:+.2f})')
    axR.fill_betweenx(logM_grid, lo, hi, color=color, alpha=0.10)

axL.set_title(r'H1: constant-$\sigma$ fits')
axR.set_title(r'H2: mass-dependent-$\sigma$ fits')
axL.set_ylabel(r'$\log_{10}(M / [M_\odot / h])$')
for ax in (axL, axR):
    ax.legend(loc='upper left', fontsize=8)

f_low_d  = 1 - det[bin_lab==0].mean()
f_high_d = 1 - det[bin_lab==1].mean()
f_tot_d  = 1 - det.mean()
plt.suptitle(fr'Hybrid selection: threshold in low bin ($f_\mathrm{{low}}$={f_low_d:.0%}), '
             fr'random {F_HIGH_HYBRID_DEMO:.0%} in high bin ($f_\mathrm{{high}}$={f_high_d:.0%}), '
             fr'$f_\mathrm{{total}}$={f_tot_d:.0%}', fontsize=11)
plt.tight_layout()
plt.show()
```

## Section 3 — Sensitivity of MOR recovery to knowledge of the selection function

### 1 — Width knowledge

$$P(\text{surv} \mid y) = \Phi\!\left(\frac{y - y_\mathrm{min}}{\sigma_\mathrm{det}}\right)$$

**Truth:** $(y_\mathrm{min}^\mathrm{true},\ \sigma_\mathrm{det}^\mathrm{true})$ both known.
**Fit:** $y_\mathrm{min}^\mathrm{model} = y_\mathrm{min}^\mathrm{true}$ (center matched), $\sigma_\mathrm{det}^\mathrm{model} = k \cdot \sigma_\mathrm{det}^\mathrm{true}$.

Scan variable: $k \in [0, \infty)$, dimensionless.

- $k = 1$: fit knows width exactly.
- $k < 1$: fit assumes a sharper completeness curve than truth; under-predicts loss in the transition tails, over-predicts loss near the limit.
- $k > 1$: fit assumes a broader completeness curve than truth; opposite direction.

Because both curves share the same center, $k \neq 1$ preserves the *total* expected loss count while distorting the *shape* of the transition zone. The MOR bias comes from the shape distortion, not the co

#### The likelihood

For the fit to model smoothness properly, its selection-aware likelihood needs to account for both the smooth curve's effect on each detected point AND the normalization over all possible outcomes. The closed form is:

$$\ln P(y_i \mid x_i, \text{det}) = \ln \phi(y_i; \mu_i, \sigma_i) + \ln \Phi\!\left(\frac{y_i - y_\mathrm{min}^\mathrm{soft}}{\sigma_\mathrm{det}^\mathrm{model}}\right) - \ln \Phi\!\left(\frac{\mu_i - y_\mathrm{min}^\mathrm{soft}}{\sqrt{\sigma_i^2 + (\sigma_\mathrm{det}^\mathrm{model})^2}}\right)$$

The normalization term uses the standard Gaussian-times-Gaussian-CDF integral (see Kelly 2007 §5.2; Mantz+10 Paper I §4.1.2). At $\sigma_\mathrm{det}^\mathrm{model} \to 0$ it reduces to the truncated Gaussian likelihood you already have — the fit believes selection is a hard step.unt mismatch.

```python
def smooth_curve_selection(y, y_min_soft, sigma_det, rng):
    """Truth-side selection: pure smooth completeness curve, no hard cut.
    P(surv | y) = Phi((y - y_min_soft) / sigma_det).
    """
    surv_prob = norm.cdf((y - y_min_soft) / sigma_det)
    return rng.uniform(0, 1, size=len(y)) < surv_prob


def neg_loglike_h2_smooth(params, x, y, y_min_soft, sigma_det_model):
    """Selection-aware H2 log-likelihood with a smooth completeness model.

    Three terms per point:
      (i)   Gaussian density of y given the MOR
      (ii)  Log survival probability of this specific point under the smooth curve
      (iii) Log total detection probability at this mass (normalization),
            given by the analytic Gaussian * Gaussian-CDF integral:
            Phi((mu - y_min_soft) / sqrt(sigma_i^2 + sigma_det_model^2))

    At sigma_det_model -> 0 this reduces to the truncated Gaussian likelihood
    (fit assumes a hard threshold). A small floor prevents numerical blow-up.
    """
    A, alpha, log_sigma0, gamma = params
    sigma0 = np.exp(log_sigma0)
    sigma_i = sigma0 * np.exp(gamma * x)
    mu = A + alpha * x
    sdm = max(sigma_det_model, 1e-4)

    log_pdf         = norm.logpdf(y, mu, sigma_i)
    log_surv_point  = norm.logcdf((y - y_min_soft) / sdm)
    combined_sigma  = np.sqrt(sigma_i ** 2 + sdm ** 2)
    log_total_surv  = norm.logcdf((mu - y_min_soft) / combined_sigma)

    return -np.sum(log_pdf + log_surv_point - log_total_surv)


def fit_h2_smooth(x, y, y_min_soft, sigma_det_model, init=(0.0, 1.5, 0.2, 0.0)):
    """Fit H2 assuming smooth completeness at sigma_det_model.
    sigma_det_model is a KNOWN input, not a fitted parameter — it represents
    how much of the true completeness curve the fit is aware of.
    """
    if len(x) < 4:
        return np.nan, np.nan, np.nan, np.nan
    x0 = [init[0], init[1], np.log(init[2]), init[3]]
    r = minimize(neg_loglike_h2_smooth, x0=x0,
                 args=(x, y, y_min_soft, sigma_det_model),
                 method='Nelder-Mead',
                 options={'xatol': 1e-5, 'fatol': 1e-5, 'maxiter': 20000})
    if not r.success:
        return np.nan, np.nan, np.nan, np.nan
    A, alpha, log_sigma0, gamma = r.x
    return A, alpha, np.exp(log_sigma0), gamma
```

```python
def scan_knowledge_ratio(sd_true, k_grid, n_realizations,
                          N_low, N_high, M_edges_low, M_edges_high,
                          y_min_soft, gamma_truth=GAMMA_SCATTER,
                          seed_offset=30000, min_detected=10):
    """For a FIXED true smoothness sd_true, scan the fit's knowledge ratio k.

    At each k, the fit assumes smooth completeness at sigma_det_model = k * sd_true.
    Truth is always smooth completeness at sd_true.
    """
    n = len(k_grid)
    keys = ['A', 'alpha', 'sigma0', 'gamma', 'f_total']
    results = {kk: np.full((n, n_realizations), np.nan) for kk in keys}
    init = (A_TRUE, ALPHA_TRUE, SIGMA_TRUE, gamma_truth)

    for i, k in enumerate(k_grid):
        sd_model = k * sd_true
        for j in range(n_realizations):
            rng = np.random.default_rng(seed_offset + i * 1000 + j)
            M, x, y, bin_lab = generate_mock(N_low, N_high,
                                             M_edges_low, M_edges_high, rng,
                                             gamma=gamma_truth)
            det = smooth_curve_selection(y, y_min_soft, sd_true, rng)
            results['f_total'][i, j] = 1 - det.mean()
            if det.sum() < min_detected:
                continue
            A, a, s0, g = fit_h2_smooth(x[det], y[det], y_min_soft, sd_model, init=init)
            # Filter obvious convergence failures — occasional at small k with large sd_true
            if np.isnan(a) or abs(a - ALPHA_TRUE) > 1.0 or abs(A) > 2.0:
                continue
            results['A'    ][i, j] = A
            results['alpha'][i, j] = a
            results['sigma0'][i, j] = s0
            results['gamma'][i, j] = g

    return results


# Log-ish spacing on k gives resolution where the threshold happens
K_GRID = np.array([0.05, 0.1, 0.15, 0.25, 0.4, 0.6, 0.8, 1.0, 1.25, 1.5, 2.0, 2.5, 3.0])
Y_MIN_SOFT = -2.1

# Scan multiple sd_true values to see how the threshold depends on the survey's smoothness.
SD_TRUE_LIST = [0.2, 0.4]
results_knowledge = {
    sd_true: scan_knowledge_ratio(sd_true, K_GRID, N_REALIZATIONS,
                                   N_LOW, N_HIGH, M_EDGES_LOW, M_EDGES_HIGH,
                                   y_min_soft=Y_MIN_SOFT)
    for sd_true in SD_TRUE_LIST
}
print(f"Knowledge scan complete for sd_true in {SD_TRUE_LIST}.")
```

```python
def loss_accounting_table(sd_true, k_grid, y_min_soft,
                          N_low, N_high, M_edges_low, M_edges_high,
                          gamma_truth=GAMMA_SCATTER, n_realizations=50,
                          seed_offset=40000):
    """Count how many lost clusters the fit's model correctly accounts for.

    For each cluster at observable y_i, computes true and model loss probabilities
    from the two Gaussian CDFs and reports expected counts of lost, model-predicted,
    and disagreeing clusters.
    """
    rows = []
    for k in k_grid:
        sd_model = max(k * sd_true, 1e-6)
        Nlost_true_acc = []
        Nlost_model_acc = []
        Ndisagree_acc = []
        for j in range(n_realizations):
            rng = np.random.default_rng(seed_offset + j)
            M, x, y, _ = generate_mock(N_low, N_high,
                                       M_edges_low, M_edges_high, rng,
                                       gamma=gamma_truth)
            p_true  = 1.0 - norm.cdf((y - y_min_soft) / sd_true)
            p_model = 1.0 - norm.cdf((y - y_min_soft) / sd_model)
            Nlost_true_acc.append(p_true.sum())
            Nlost_model_acc.append(p_model.sum())
            Ndisagree_acc.append(np.abs(p_true - p_model).sum())
        Nlt = np.mean(Nlost_true_acc)
        Nlm = np.mean(Nlost_model_acc)
        Nd  = np.mean(Ndisagree_acc)
        frac_correct = 1.0 - Nd / Nlt if Nlt > 0 else np.nan
        rows.append({
            'k':                   k,
            'sigma_det_model':     sd_model,
            'N_lost_true':         Nlt,
            'N_lost_model':        Nlm,
            'N_disagree':          Nd,
            'frac_correctly_modeled': frac_correct,
        })
    return rows


def print_loss_table(rows, sd_true):
    print(f"\nLoss accounting for sigma_det_true = {sd_true}")
    print(f"  (averaged over 50 realisations, N = {N_LOW + N_HIGH} clusters per mock)")
    print(f"{'k':>6} {'sd_model':>10} {'N_lost_true':>13} {'N_lost_model':>14} "
          f"{'N_disagree':>12} {'% correct':>11}")
    for r in rows:
        print(f"{r['k']:6.2f} {r['sigma_det_model']:10.3f} "
              f"{r['N_lost_true']:13.1f} {r['N_lost_model']:14.1f} "
              f"{r['N_disagree']:12.1f} {r['frac_correctly_modeled']*100:10.1f}%")


for sd_true in SD_TRUE_LIST:
    rows = loss_accounting_table(sd_true, K_GRID, Y_MIN_SOFT,
                                  N_LOW, N_HIGH, M_EDGES_LOW, M_EDGES_HIGH)
    print_loss_table(rows, sd_true)
```

```python
def plot_knowledge_scan_with_loss_frac(results_dict, k_grid, sd_true_list,
                                        y_min_soft, tolerances=None):
    """Same as plot_knowledge_scan but adds a secondary x-axis showing
    fraction of losses correctly modeled at each k.
    """
    if tolerances is None:
        tolerances = {'alpha': 0.03, 'sigma0': 0.02, 'gamma': 0.05}
    colors = ['navy', 'crimson', 'darkgreen']
    markers = ['o', 's', 'D']

    fig, axes = plt.subplots(1, 3, figsize=(15, 4.8))

    # Compute fraction-correct for each sd_true and each k
    frac_correct_by_sd = {}
    for sd_true in sd_true_list:
        rows = loss_accounting_table(sd_true, k_grid, y_min_soft,
                                     N_LOW, N_HIGH, M_EDGES_LOW, M_EDGES_HIGH,
                                     n_realizations=30)
        frac_correct_by_sd[sd_true] = np.array([r['frac_correctly_modeled']
                                                for r in rows])

    for ax, key, truth, ylabel, title in [
        (axes[0], 'alpha',  ALPHA_TRUE,     r'Fitted $\alpha$',     r'Slope $\alpha$'),
        (axes[1], 'sigma0', SIGMA_TRUE,     r'Fitted $\sigma_0$',   r'Scatter $\sigma_0$'),
        (axes[2], 'gamma',  GAMMA_SCATTER,  r'Fitted $\gamma$',     r'Mass-dep $\gamma$'),
    ]:
        tol = tolerances[key]
        ax.axhspan(truth - tol, truth + tol, color='goldenrod', alpha=0.10,
                   label=fr'$\pm${tol} tolerance')
        ax.axhline(truth, color='goldenrod', ls=':', lw=2, label=f'Truth ({truth})')
        ax.axvline(1.0, color='gray', ls='--', lw=1, alpha=0.6, label='k=1')

        for (sd_true, r), color, mk in zip(results_dict.items(), colors, markers):
            d_med = np.nanmedian(r[key], axis=1)
            d_lo  = np.nanpercentile(r[key], 16, axis=1)
            d_hi  = np.nanpercentile(r[key], 84, axis=1)
            ax.plot(k_grid, d_med, marker=mk, ls='-', color=color, lw=2,
                    label=fr'$\sigma_\mathrm{{det}}^\mathrm{{true}}$={sd_true}')
            ax.fill_between(k_grid, d_lo, d_hi, color=color, alpha=0.15)

        ax.set_xlabel(r'$k = \sigma_\mathrm{det}^\mathrm{model} / \sigma_\mathrm{det}^\mathrm{true}$')
        ax.set_ylabel(ylabel)
        ax.set_title(title)
        ax.legend(fontsize=7, loc='best')

        # Secondary x-axis showing fraction correctly modeled (use first sd_true in list)
        first_sd = list(results_dict.keys())[0]
        fc = frac_correct_by_sd[first_sd]
        ax_top = ax.twiny()
        ax_top.set_xlim(ax.get_xlim())
        # Tick locations at select k values
        tick_ks = [k_grid[0], k_grid[len(k_grid)//4], k_grid[len(k_grid)//2],
                   k_grid[3*len(k_grid)//4], k_grid[-1]]
        tick_labels = [f'{np.interp(tk, k_grid, fc)*100:.0f}%' for tk in tick_ks]
        ax_top.set_xticks(tick_ks)
        ax_top.set_xticklabels(tick_labels, fontsize=8)
        ax_top.set_xlabel(fr'% losses correctly modeled ($\sigma_\mathrm{{det}}^\mathrm{{true}}$={first_sd})',
                          fontsize=8)

    plt.tight_layout()
    plt.show()


plot_knowledge_scan_with_loss_frac(results_knowledge, K_GRID,
                                    SD_TRUE_LIST, Y_MIN_SOFT)
```

### 2 — Center-value knowledge

$$P(\text{surv} \mid y) = \Phi\!\left(\frac{y - y_\mathrm{min}}{\sigma_\mathrm{det}}\right)$$

**Truth:** $y_\mathrm{min}^\mathrm{true}$ known, $\sigma_\mathrm{det}^\mathrm{true}$ known.
**Fit:** $\sigma_\mathrm{det}^\mathrm{model} = \sigma_\mathrm{det}^\mathrm{true}$ (width known), $y_\mathrm{min}^\mathrm{model} \ne y_\mathrm{min}^\mathrm{true}$ (center wrong).

Scan variable: $\Delta = y_\mathrm{min}^\mathrm{model} - y_\mathrm{min}^\mathrm{true}$, in units of $\ln(\mathrm{Obs})$.
$\Delta = 0$: fit knows center exactly.
$\Delta > 0$: fit places the limit above truth (over-predicts loss).
$\Delta < 0$: fit places the limit below truth (under-predicts loss).

### Scan function

```python
def scan_center_knowledge(delta_grid, n_realizations,
                          N_low, N_high, M_edges_low, M_edges_high,
                          y_min_soft_true, sigma_det_true,
                          gamma_truth=GAMMA_SCATTER,
                          seed_offset=50000, min_detected=10):
    """Fix width (fit uses sigma_det_true). Scan center offset Delta.
    Fit uses y_min_soft_model = y_min_soft_true + Delta as its assumed centre.
    """
    n = len(delta_grid)
    keys = ['A', 'alpha', 'sigma0', 'gamma', 'f_total']
    results = {kk: np.full((n, n_realizations), np.nan) for kk in keys}
    init = (A_TRUE, ALPHA_TRUE, SIGMA_TRUE, gamma_truth)

    for i, delta in enumerate(delta_grid):
        y_min_model = y_min_soft_true + delta
        for j in range(n_realizations):
            rng = np.random.default_rng(seed_offset + i * 1000 + j)
            M, x, y, bin_lab = generate_mock(N_low, N_high,
                                             M_edges_low, M_edges_high, rng,
                                             gamma=gamma_truth)
            det = smooth_curve_selection(y, y_min_soft_true, sigma_det_true, rng)
            results['f_total'][i, j] = 1 - det.mean()
            if det.sum() < min_detected:
                continue
            A, a, s0, g = fit_h2_smooth(x[det], y[det],
                                         y_min_model, sigma_det_true, init=init)
            if np.isnan(a) or abs(a - ALPHA_TRUE) > 1.0 or abs(A) > 2.0:
                continue
            results['A'][i, j]      = A
            results['alpha'][i, j]  = a
            results['sigma0'][i, j] = s0
            results['gamma'][i, j]  = g
    return results
```

### Run the scan

```python
DELTA_GRID = np.linspace(-0.5, 0.5, 11)
Y_MIN_SOFT_TRUE = -2.1

results_center = {
    sd_true: scan_center_knowledge(
        DELTA_GRID, N_REALIZATIONS,
        N_LOW, N_HIGH, M_EDGES_LOW, M_EDGES_HIGH,
        y_min_soft_true=Y_MIN_SOFT_TRUE,
        sigma_det_true=sd_true,
    )
    for sd_true in SD_TRUE_LIST
}
print(f"Center-knowledge scan complete for sd_true in {SD_TRUE_LIST}.")
```

```python
def loss_accounting_table_center(y_min_soft_true, sigma_det_true, delta_grid,
                                  N_low, N_high, M_edges_low, M_edges_high,
                                  gamma_truth=GAMMA_SCATTER, n_realizations=50,
                                  seed_offset=60000):
    """Loss accounting for the center-knowledge scan.
    Both true and model curves use sigma_det_true (width matched); only center differs.
    """
    rows = []
    for delta in delta_grid:
        y_min_model = y_min_soft_true + delta
        Nlost_true_acc = []
        Nlost_model_acc = []
        Ndisagree_acc = []
        for j in range(n_realizations):
            rng = np.random.default_rng(seed_offset + j)
            M, x, y, _ = generate_mock(N_low, N_high,
                                       M_edges_low, M_edges_high, rng,
                                       gamma=gamma_truth)
            p_true  = 1.0 - norm.cdf((y - y_min_soft_true)  / sigma_det_true)
            p_model = 1.0 - norm.cdf((y - y_min_model)      / sigma_det_true)
            Nlost_true_acc.append(p_true.sum())
            Nlost_model_acc.append(p_model.sum())
            Ndisagree_acc.append(np.abs(p_true - p_model).sum())
        Nlt = np.mean(Nlost_true_acc)
        Nlm = np.mean(Nlost_model_acc)
        Nd  = np.mean(Ndisagree_acc)
        frac_correct = 1.0 - Nd / Nlt if Nlt > 0 else np.nan
        rows.append({
            'delta':                delta,
            'y_min_model':          y_min_model,
            'N_lost_true':          Nlt,
            'N_lost_model':         Nlm,
            'N_disagree':           Nd,
            'frac_correctly_modeled': frac_correct,
        })
    return rows


def print_loss_table_center(rows, sd_true):
    print(f"\nLoss accounting for sigma_det_true = {sd_true}, "
          f"center scan (width matched)")
    print(f"  (averaged over 50 realisations, N = {N_LOW + N_HIGH} clusters per mock)")
    print(f"{'delta':>7} {'y_min_model':>13} {'N_lost_true':>13} "
          f"{'N_lost_model':>14} {'N_disagree':>12} {'% correct':>11}")
    for r in rows:
        print(f"{r['delta']:+7.2f} {r['y_min_model']:13.3f} "
              f"{r['N_lost_true']:13.1f} {r['N_lost_model']:14.1f} "
              f"{r['N_disagree']:12.1f} {r['frac_correctly_modeled']*100:10.1f}%")


for sd_true in SD_TRUE_LIST:
    rows = loss_accounting_table_center(
        Y_MIN_SOFT_TRUE, sd_true, DELTA_GRID,
        N_LOW, N_HIGH, M_EDGES_LOW, M_EDGES_HIGH,
    )
    print_loss_table_center(rows, sd_true)
```

### Plot

```python
def plot_center_scan_with_loss_frac(results_dict, delta_grid, sd_true_list,
                                     y_min_soft_true, tolerances=None):
    if tolerances is None:
        tolerances = {'alpha': 0.03, 'sigma0': 0.02, 'gamma': 0.05}
    colors = ['navy', 'crimson', 'darkgreen']
    markers = ['o', 's', 'D']

    fig, axes = plt.subplots(1, 3, figsize=(15, 4.8))

    # % correctly modeled per delta, for the first sd_true (used for the top axis)
    first_sd = sd_true_list[0]
    rows_top = loss_accounting_table_center(
        y_min_soft_true, first_sd, delta_grid,
        N_LOW, N_HIGH, M_EDGES_LOW, M_EDGES_HIGH,
        n_realizations=30,
    )
    frac_correct = np.array([r['frac_correctly_modeled'] for r in rows_top])

    for ax, key, truth, ylabel, title in [
        (axes[0], 'alpha',  ALPHA_TRUE,    r'Fitted $\alpha$',   r'Slope $\alpha$'),
        (axes[1], 'sigma0', SIGMA_TRUE,    r'Fitted $\sigma_0$', r'Scatter $\sigma_0$'),
        (axes[2], 'gamma',  GAMMA_SCATTER, r'Fitted $\gamma$',   r'Mass-dep $\gamma$'),
    ]:
        tol = tolerances[key]
        ax.axhspan(truth - tol, truth + tol, color='goldenrod', alpha=0.10,
                   label=fr'$\pm${tol} tolerance')
        ax.axhline(truth, color='goldenrod', ls=':', lw=2, label=f'Truth ({truth})')
        ax.axvline(0.0, color='gray', ls='--', lw=1, alpha=0.6,
                   label=r'$\Delta$=0 (full knowledge)')

        for (sd_true, r), color, mk in zip(results_dict.items(), colors, markers):
            d_med = np.nanmedian(r[key], axis=1)
            d_lo  = np.nanpercentile(r[key], 16, axis=1)
            d_hi  = np.nanpercentile(r[key], 84, axis=1)
            ax.plot(delta_grid, d_med, marker=mk, ls='-', color=color, lw=2,
                    label=fr'$\sigma_\mathrm{{det}}^\mathrm{{true}}$={sd_true}')
            ax.fill_between(delta_grid, d_lo, d_hi, color=color, alpha=0.15)

        ax.set_xlabel(r'$\Delta = y_\mathrm{min}^\mathrm{model} - y_\mathrm{min}^\mathrm{true}$')
        ax.set_ylabel(ylabel)
        ax.set_title(title)
        ax.legend(fontsize=7, loc='best')

        # Top axis showing % correctly modeled
        ax_top = ax.twiny()
        ax_top.set_xlim(ax.get_xlim())
        tick_ds = [delta_grid[0], delta_grid[len(delta_grid)//4],
                   delta_grid[len(delta_grid)//2],
                   delta_grid[3*len(delta_grid)//4], delta_grid[-1]]
        tick_labels = [f'{np.interp(td, delta_grid, frac_correct)*100:.0f}%'
                       for td in tick_ds]
        ax_top.set_xticks(tick_ds)
        ax_top.set_xticklabels(tick_labels, fontsize=8)
        ax_top.set_xlabel(
            fr'% losses correctly modeled ($\sigma_\mathrm{{det}}^\mathrm{{true}}$={first_sd})',
            fontsize=8,
        )

    plt.tight_layout()
    plt.show()


plot_center_scan_with_loss_frac(results_center, DELTA_GRID,
                                 SD_TRUE_LIST, Y_MIN_SOFT_TRUE)
```

### Threshold extraction

```python
def find_delta_threshold(results, delta_grid, key, truth, tolerance):
    """Delta range for which |median(fit) - truth| stays within tolerance.
    Returns (delta_lower, delta_upper) or (nan, nan) if the fit never enters tolerance.
    """
    d_med = np.nanmedian(results[key], axis=1)
    within = np.abs(d_med - truth) <= tolerance
    if not within.any():
        return np.nan, np.nan
    indices = np.where(within)[0]
    return delta_grid[indices.min()], delta_grid[indices.max()]


print("\nAllowed range of Delta for each parameter within tolerance:")
print(f"{'sd_true':>8} {'d_alpha':>18} {'d_sigma0':>18} {'d_gamma':>18}")
for sd_true, r in results_center.items():
    da  = find_delta_threshold(r, DELTA_GRID, 'alpha',  ALPHA_TRUE,    0.03)
    ds  = find_delta_threshold(r, DELTA_GRID, 'sigma0', SIGMA_TRUE,    0.02)
    dg  = find_delta_threshold(r, DELTA_GRID, 'gamma',  GAMMA_SCATTER, 0.05)
    print(f"{sd_true:8.2f} [{da[0]:+.2f},{da[1]:+.2f}] "
          f"[{ds[0]:+.2f},{ds[1]:+.2f}] [{dg[0]:+.2f},{dg[1]:+.2f}]")
```
